**Perceptron**

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sklearn.linear_model import Perceptron
from sklearn.datasets import load_breast_cancer
from mlxtend.plotting import plot_decision_regions
```

```python
data= load_breast_cancer()
```

```python
df= pd.DataFrame(data.data , columns=data.feature_names)
df['target']=data.target
```

```python
X=df.iloc[:,[0,1]]
```

```python
y=df['target']
```

```python
p=Perceptron()
```

```python
p.fit(X,y)
```

```python
print('Model Coefficient:',p.coef_)
print('Model Intercept:',p.intercept_)
```

```python
z=p.score(X,y)
print('Accuracy Score:',z)
```

```python
from keras.models import Sequential
from keras.layers import Dense, Activation
import numpy as np
import pandas as pd
from sklearn import datasets
```

```python
iris= datasets.load_iris()
X,y= datasets.load_iris(return_X_y=True)
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test= train_test_split(X,y,test_size=0.40)
```

```python
model = Sequential()
model.add(Dense(2,input_shape=(4,)))
model.add(Activation('sigmoid'))
model.add(Dense(1))
model.add(Activation('sigmoid'))
```

```python
model.compile(loss='mean_squared_error',optimizer='sgd',metrics=['accuracy'])
```

```python
model.summary()
model.fit(X_train,y_train,epochs=10)
```

```python
score= model.evaluate(X_test,y_test)
print('Loss:',score[0])
print('Accuracy:',score[1])
```

```python
import tensorflow as tf
from tensorflow import keras
from keras.models import Sequential
from keras import Input
from keras.layers import Dense
import pandas as pd
import numpy as np
import sklearn
from sklearn.metrics import classification_report
import matplotlib
import matplotlib.pyplot as plt
```

```python
#loading the digits data
(X_train, y_train), (X_test, y_test) = keras.datasets.mnist.load_data()
```

```python
print('shape of X_train', X_train.shape)
print('shape of y_train', y_train.shape)
print('shape of X_test', X_test.shape)
print('shape of y_test', y_test.shape)
```

```python
#display image of the first 10 digits
fig, axs = plt.subplots(2, 5, sharey=False, tight_layout=True, figsize=(12,6), facecolor='white')
n=0
for i in range(0,2):
  for j in range(0,5):
    axs[i,j].matshow(X_train[n])
    axs[i,j].set(title=y_train[n])
    n=n+1
plt.show()
```

```python
#reshape and normalize input data
X_train= X_train.reshape(60000,784).astype('float32') /255
X_test= X_test.reshape(10000,784).astype('float32') /255
```

```python
print('shape of x_train', X_train.shape)
print('shape of y_train', y_train.shape)
```

```python
#design the deep ff nn architecture
model= Sequential(name= 'DFF-model')
model.add(Input(shape=(784,), name='input_layer'))
model.add(Dense(128, activation='relu', name='hidden_layer_1',kernel_initializer='he_normal'))
model.add(Dense(64, activation='relu', name='hidden_layer_2',kernel_initializer='he_normal'))
model.add(Dense(32, activation='softmax', name='hidden_layer_3',kernel_initializer='he_normal'))
model.add(Dense(10, activation='softmax', name='output_layer'))
```

```python
#compare keras model
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.summary()
```

```python
#fit keras model on data
model.fit(X_train, y_train, batch_size=10, epochs=5, verbose=1, validation_split=0.2, shuffle=True)
```

```python
#predict class labels
pred_labels= model.predict(X_test)
pred_labels_te= np.argmax(pred_labels, axis=1)
```

```python
#predict class labels
pred_labels= model.predict(X_train)
pred_labels_tr= np.argmax(pred_labels, axis=1)
```

```python
#model performance summary and evaluation
print('')
print('-----------model summary-----------------')
model.summary()
print('')
```

```python
print('-----------model evaluation-----------------')
print(classification_report(y_train, pred_labels_tr))
print('')
```

```python
print('-----------model evaluation-----------------')
print(classification_report(y_test, pred_labels_te))
print('')
```

```python
import matplotlib.pyplot as plt
from keras.models import Sequential
from keras.layers import Dense, Input

# Helper function to create the same model architecture
def create_model():
    model = Sequential(name='DFF-model')
    model.add(Input(shape=(784,), name='input_layer'))
    model.add(Dense(128, activation='relu', name='hidden_layer_1', kernel_initializer='he_normal'))
    model.add(Dense(64, activation='relu', name='hidden_layer_2', kernel_initializer='he_normal'))
    model.add(Dense(32, activation='softmax', name='hidden_layer_3', kernel_initializer='he_normal'))
    model.add(Dense(10, activation='softmax', name='output_layer'))
    return model

# 1. Train model with Adam optimizer
model_adam = create_model()
model_adam.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
history_adam = model_adam.fit(X_train, y_train, batch_size=10, epochs=5, validation_split=0.2, shuffle=True, verbose=0)

# 2. Train model with SGD optimizer
model_sgd = create_model()
model_sgd.compile(optimizer='sgd', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
history_sgd = model_sgd.fit(X_train, y_train, batch_size=10, epochs=5, validation_split=0.2, shuffle=True, verbose=0)

# 3. Plot Accuracy and Loss comparison
epochs = range(1, 6)
fig, axs = plt.subplots(1, 2, figsize=(14, 5))

# Accuracy comparison
axs[0].plot(epochs, history_adam.history['accuracy'], 'b-', label='Adam Train Acc')
axs[0].plot(epochs, history_adam.history['val_accuracy'], 'b--', label='Adam Val Acc')
axs[0].plot(epochs, history_sgd.history['accuracy'], 'r-', label='SGD Train Acc')
axs[0].plot(epochs, history_sgd.history['val_accuracy'], 'r--', label='SGD Val Acc')
axs[0].set_title('Accuracy Comparison')
axs[0].set_xlabel('Epochs')
axs[0].set_ylabel('Accuracy')
axs[0].legend()
axs[0].grid(True)

# Loss comparison
axs[1].plot(epochs, history_adam.history['loss'], 'b-', label='Adam Train Loss')
axs[1].plot(epochs, history_adam.history['val_loss'], 'b--', label='Adam Val Loss')
axs[1].plot(epochs, history_sgd.history['loss'], 'r-', label='SGD Train Loss')
axs[1].plot(epochs, history_sgd.history['val_loss'], 'r--', label='SGD Val Loss')
axs[1].set_title('Loss Comparison')
axs[1].set_xlabel('Epochs')
axs[1].set_ylabel('Loss')
axs[1].legend()
axs[1].grid(True)

plt.tight_layout()
plt.show()
```
